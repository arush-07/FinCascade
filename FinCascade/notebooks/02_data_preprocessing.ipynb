```python
prices = pd.read_csv(
    DATA_DIR / "market_prices_raw.csv",
    index_col=0,
    parse_dates=True
)

metadata = pd.read_csv(
    DATA_DIR / "entities.csv"
)

prices = prices.sort_index()

print("Prices shape:", prices.shape)
print("Entities:", len(metadata))
print("Date range:", prices.index.min(), "to", prices.index.max())

prices.head()
```

```python
# Use valid NIFTY 50 trading dates as the master calendar
nifty_dates = prices["^NSEI"].dropna().index

aligned_prices = prices.reindex(nifty_dates)

print("Original shape:", prices.shape)
print("Aligned shape :", aligned_prices.shape)

print("\nTop missing-value counts after alignment:")
print(
    aligned_prices.isna()
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
# Forward-fill only short gaps caused by different market holidays
clean_prices = aligned_prices.ffill(limit=2)

# Drop any rows that still contain missing values
clean_prices = clean_prices.dropna().copy()

print("Aligned shape :", aligned_prices.shape)
print("Clean shape   :", clean_prices.shape)

print("\nRemaining missing values:")
print(clean_prices.isna().sum().sum())

clean_prices.to_csv(
    DATA_DIR / "market_prices_clean.csv"
)

print("\nSaved: market_prices_clean.csv")
```

```python
returns = clean_prices.pct_change().dropna()

# Known Tata Motors demerger adjustment date
demerger_date = pd.Timestamp("2025-10-14")

if demerger_date in returns.index:
    returns.loc[demerger_date, "TMPV.NS"] = np.nan

# Model-ready dataset: remove rows containing any NaN
returns_model = returns.dropna().copy()

print("Returns shape       :", returns.shape)
print("Model-ready shape   :", returns_model.shape)

print("\nMissing values:")
print(returns_model.isna().sum().sum())

print("\nInfinite values:")
print(np.isinf(returns_model).sum().sum())

print("\nReturn range:")
print("Minimum:", returns_model.min().min())
print("Maximum:", returns_model.max().max())
```

```python
returns.to_csv(
    DATA_DIR / "market_returns_clean.csv"
)

returns_model.to_csv(
    DATA_DIR / "market_returns_model.csv"
)

print("Saved:")
print("- market_returns_clean.csv")
print("- market_returns_model.csv")

print("\nModel dataset shape:", returns_model.shape)
```

```python
asset_stats = pd.DataFrame({
    "mean_daily_return": returns_model.mean(),
    "daily_volatility": returns_model.std(),
    "annualized_return": returns_model.mean() * 252,
    "annualized_volatility": returns_model.std() * np.sqrt(252),
    "min_return": returns_model.min(),
    "max_return": returns_model.max()
})

print("Asset statistics created.")
print("Shape:", asset_stats.shape)

asset_stats.head()
```

```python
meta_lookup = metadata.set_index("ticker")

asset_stats["name"] = asset_stats.index.map(
    meta_lookup["name"]
)

asset_stats["type"] = asset_stats.index.map(
    meta_lookup["type"]
)

asset_stats["sector"] = asset_stats.index.map(
    meta_lookup["sector"]
)

asset_stats = asset_stats[
    [
        "name",
        "type",
        "sector",
        "mean_daily_return",
        "annualized_return",
        "daily_volatility",
        "annualized_volatility",
        "min_return",
        "max_return"
    ]
]

asset_stats.head()
```

```python
volatility_table = asset_stats.sort_values(
    "annualized_volatility",
    ascending=False
)[
    [
        "name",
        "sector",
        "annualized_volatility"
    ]
].head(10).copy()

# Convert to percentage
volatility_table["annualized_volatility"] *= 100

volatility_table
```

```python
asset_stats.to_csv(
    DATA_DIR / "asset_statistics.csv"
)

print("Saved: asset_statistics.csv")
print("Assets:", len(asset_stats))
```
